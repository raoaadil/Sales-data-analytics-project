```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading Data
```

```python
df = pd.read_csv('messy_ecommerce_sales_dataset.csv')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
# Data Cleaning
```

```python
df.loc[df['Age'] < 0, 'Age'] = np.nan
```

```python
df.describe()
```

```python
df['Sales'].plot.box()
```

```python
upper = df['Sales'].quantile(0.99)
df['Sales'] = df['Sales'].clip(upper=upper)
```

```python
df['Sales'].plot.box()
```

```python
df.describe()
```

```python
df.describe(include='str')
```

```python
df['Order_Date']
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed')
```

```python
df['Order_Date'].isnull().sum()
```

```python
df
```

```python
df['Gender'].unique()
```

```python
df['City'].unique()
```

```python
df['City'] = df['City'].replace({'bangalore': 'Bangalore', 'mumbai':'Mumbai', 'delhi':'Delhi', 'hyderabad':'Hyderabad', 'pune':'Pune', 'chennai':'Chennai', 'kolkata':'Kolkata'})
```

```python
df['City'].unique()
```

```python
df.columns
```

```python
df['Category'].unique()
```

```python
df['Category'] = df['Category'].str.strip()
```

```python
df['Category'].unique()
```

```python
df['Product'].unique()
```

```python
df['Payment_Method'].unique()
```

```python
df['Payment_Method'] = df['Payment_Method'].replace({'DEBIT CARD':'Debit Card', 'CREDIT CARD':'Credit Card', 'CASH ON DELIVERY':'Cash on Delivery', 'NET BANKING':'Net Banking'})
```

```python
df['Payment_Method'].unique()
```

```python
df['Returned'].unique()
```

```python
df['year'] = df['Order_Date'].dt.year
df['month'] = df['Order_Date'].dt.month
df['day'] = df['Order_Date'].dt.day
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Discount'] = df['Discount'].fillna(df['Discount'].mean())
df['Profit'] = df['Profit'].fillna(df['Profit'].mean())
df['City'] = df['City'].fillna(df['City'].mode()[0])
```

```python
df = df.dropna(subset=['Customer_Name', 'Gender'])
```

```python
df.info()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df.to_csv('cleaned_data.csv', index=False)
```

```python
# VISUALISATION 
```

```python
Category_sales = df.groupby('Category')['Sales'].sum()
```

```python
plt.figure(figsize=(5,5))
Category_sales.plot(kind='bar')
plt.title('Category wise Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
top_customers = df.groupby('Customer_ID')['Sales'].sum().sort_values(ascending=False).head(10)
top_customers
```

```python
plt.figure(figsize=(5,5))
top_customers.plot(kind='bar')
plt.xlabel('Customer IDs')
plt.ylabel('Sales')
```

```python
profit_earn = df.groupby('City')['Profit'].sum()
```

```python
plt.figure(figsize = (5, 5))
profit_earn.plot(kind='barh')
plt.xlabel('Profit')
plt.ylabel('City')
plt.show()
```

```python
city_sales = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(5)
```

```python
plt.figure(figsize=(5, 5))
city_sales.plot(kind='bar')
```

```python
most_sold_products = df.groupby('Product')['Quantity'].sum()
plt.figure(figsize=(8, 8))
most_sold_products.plot(kind='bar')
```

```python
sales_trend = df.groupby(['year', 'month'])['Sales'].sum()
```

```python
plt.figure(figsize=(12, 5))
sales_trend.plot(kind='line', marker='o')
plt.xlabel('Year-Month')
plt.ylabel('Sales')
```

```python
number_of_returns = df[df['Returned'] == 'Yes'].groupby('Category')['Returned'].count()
```

```python
plt.figure(figsize=(5,5))
number_of_returns.plot(kind='bar')
plt.ylabel('Number of returns')
```
